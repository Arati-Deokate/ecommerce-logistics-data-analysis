```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("Date conversion completed!")
```

```python
delivered = orders[
    orders["order_status"] == "delivered"
].copy()

print("Total delivered orders:", len(delivered))
print(
    "Delivered orders with actual delivery date:",
    delivered["order_delivered_customer_date"].notna().sum()
)
print(
    "Delivered orders missing delivery date:",
    delivered["order_delivered_customer_date"].isna().sum()
)
```

```python
delivered[
    delivered["order_delivered_customer_date"].isna()
]
```

```python
delivered["delivery_time_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
print(delivered["delivery_time_days"].describe())
```

```python
delivered["delivery_delay_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
print(delivered["delivery_delay_days"].describe())
```

```python
on_time = (
    delivered["delivery_delay_days"] <= 0
).sum()

late = (
    delivered["delivery_delay_days"] > 0
).sum()

total_with_delay = delivered["delivery_delay_days"].notna().sum()

on_time_rate = (on_time / total_with_delay) * 100
late_rate = (late / total_with_delay) * 100

print("Orders used for delay calculation:", total_with_delay)
print("On-Time/Early Orders:", on_time)
print("Late Orders:", late)
print(f"On-Time Delivery Rate: {on_time_rate:.2f}%")
print(f"Late Delivery Rate: {late_rate:.2f}%")
```

```python
Q1 = delivered["delivery_time_days"].quantile(0.25)
Q3 = delivered["delivery_time_days"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
```

```python
outliers = delivered[
    (delivered["delivery_time_days"] < lower_bound) |
    (delivered["delivery_time_days"] > upper_bound)
]

print(
    "Potential delivery-time outliers:",
    len(outliers)
)
```

```python
# FAST TASK 2 PREPROCESSING

import pandas as pd
import numpy as np

# Work only with delivered orders having actual delivery dates
task2_data = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].notna())
].copy()

# Delivery time
task2_data["delivery_time_days"] = (
    task2_data["order_delivered_customer_date"]
    - task2_data["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# Delivery delay
task2_data["delivery_delay_days"] = (
    task2_data["order_delivered_customer_date"]
    - task2_data["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Early / On-Time / Late classification
task2_data["delivery_performance"] = np.where(
    task2_data["delivery_delay_days"] < 0,
    "Early",
    np.where(
        task2_data["delivery_delay_days"] == 0,
        "On Time",
        "Late"
    )
)

# Late delivery flag
task2_data["late_delivery"] = np.where(
    task2_data["delivery_delay_days"] > 0,
    1,
    0
)

# IQR outlier flag
Q1 = task2_data["delivery_time_days"].quantile(0.25)
Q3 = task2_data["delivery_time_days"].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

task2_data["delivery_time_outlier"] = np.where(
    task2_data["delivery_time_days"] > upper_bound,
    1,
    0
)

# Save final dataset
task2_data.to_csv(
    "Olist_Task2_Preprocessed_Data.csv",
    index=False
)

# Final summary
print("===== TASK 2 COMPLETE =====")
print("Original records:", len(orders))
print("Final analysis records:", len(task2_data))
print("Average delivery time:",
      round(task2_data["delivery_time_days"].mean(), 2), "days")
print("Average delivery delay:",
      round(task2_data["delivery_delay_days"].mean(), 2), "days")
print("Late orders:",
      task2_data["late_delivery"].sum())
print("Potential outliers:",
      task2_data["delivery_time_outlier"].sum())
print("File saved: Olist_Task2_Preprocessed_Data.csv")
```
